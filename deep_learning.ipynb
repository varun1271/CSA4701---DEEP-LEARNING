```python
import numpy as np
import matplotlib.pyplot as plt
A = np.random.randint(1, 10, (6, 5)); B = np.random.randint(1, 10, (5, 4))
C = A @ B
print("Dot Product:", np.dot(A[0], A[1]))
print("Transpose:\n", A.T)
print("Element Multiplication:\n", A*A)
print("Cosine Similarity:\n", (A @ A.T) / np.outer(np.linalg.norm(A,axis=1), np.linalg.norm(A,axis=1)))
plt.imshow(C); plt.colorbar(); plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
X = np.random.rand(200, 5)
X = (X - X.mean(axis=0)) / X.std(axis=0)
C = np.cov(X.T)
values, vectors = np.linalg.eigh(C)
print("Eigenvalues:", values)
print("Variance:", values / sum(values) * 100)
sns.heatmap(np.corrcoef(X.T), annot=True); plt.show()
```

```python
import numpy as np
from scipy.stats import entropy
import matplotlib.pyplot as plt
P = np.array([0.5, 0.3, 0.2])
Q = np.array([0.4, 0.4, 0.2])
print("Entropy:", entropy(P))
print("Cross Entropy:", -np.sum(P*np.log(Q)))
print("KL Divergence:", entropy(P, Q))
p = np.linspace(0.01, 0.99, 100)
plt.plot(p, -np.log(p)); plt.xlabel("Probability"); plt.ylabel("Loss"); plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
X = np.linspace(1, 10, 150).reshape(-1, 1); y = X.ravel()**2 + np.random.randn(150)*5
for d in [1, 3, 9, 15]:
    P = PolynomialFeatures(d).fit_transform(X)
    model = LinearRegression().fit(P, y)
    print("Degree:", d, "Error:", np.mean((y-model.predict(P))**2))
    plt.plot(X, model.predict(P), label=str(d))
plt.scatter(X, y); plt.legend(); plt.show()
```

```python
import numpy as np
x = 3.0
f = lambda x: x**2 + 2*x
h = 0.001
analytical = 2*x + 2
forward = (f(x+h)-f(x))/h
central = (f(x+h)-f(x-h))/(2*h)
print("Analytical:", analytical)
print("Forward:", forward)
print("Central:", central)
```

```python
import numpy as np
import matplotlib.pyplot as plt
epochs = np.arange(1,101)
fixed = np.full(100,0.1)
step = np.where(epochs<50,0.1,0.01)
exp = 0.1*np.exp(-0.03*epochs)
cos = 0.05*(1+np.cos(np.pi*epochs/100))
plt.plot(epochs,fixed,label="Fixed")
plt.plot(epochs,step,label="Step"); plt.plot(epochs,exp,label="Exponential"); plt.plot(epochs,cos,label="Cosine")
plt.legend(); plt.show()
```

```python
import tensorflow as tf
import numpy as np
X = np.random.rand(200,5)
y = (X.sum(axis=1)>2.5).astype(int)
model = tf.keras.Sequential([tf.keras.layers.Dense(16,activation='relu'),tf.keras.layers.Dense(8,activation='relu'),tf.keras.layers.Dense(1,activation='sigmoid')])
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
model.fit(X,y,epochs=5)
print("Accuracy:",model.evaluate(X,y)[1])
```

```python
import tensorflow as tf
import numpy as np
X = np.random.rand(500,5); y = (X.sum(axis=1)>2.5).astype(int)
for rate in [0,0.2,0.5]:
    model = tf.keras.Sequential([tf.keras.layers.Dense(32,activation='relu',kernel_regularizer=tf.keras.regularizers.l2(0.01)),tf.keras.layers.Dropout(rate),tf.keras.layers.Dense(1,activation='sigmoid')])
    model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
    model.fit(X,y,epochs=10,validation_split=0.2,callbacks=[tf.keras.callbacks.EarlyStopping(patience=2)],verbose=0)
    print("Dropout:",rate,"Accuracy:",model.evaluate(X,y,verbose=0)[1])
```
